```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

# ============================================================
# LAB 01 - AULA 03
# Pré-processamento, Stopwords e N-grams
# ============================================================

# Dataset de Atendimento Financeiro
dados = {
    'mensagem': [
        'Como posso emitir a segunda via do meu boleto?',
        'Preciso da 2a via da minha fatura atrasada',
        'Quero negociar o pagamento da minha dívida',
        'Como fazer um acordo para pagar o débito em aberto?',
        'Gostaria de alterar meu endereço de cadastramento',
        'Onde atualizo os meus dados residenciais no app?'
    ],

    'intencao': [
        'segunda_via',
        'segunda_via',
        'negociar_divida',
        'negociar_divida',
        'atualizar_cadastro',
        'atualizar_cadastro'
    ]
}

# Criando o DataFrame
df1 = pd.DataFrame(dados)

# ============================================================
# STOPWORDS PERSONALIZADAS EM PORTUGUÊS
# ============================================================

stopwords_pt = [
    'de', 'da', 'do', 'dos', 'das',
    'a', 'o', 'as', 'os',
    'em', 'para',
    'com', 'por',
    'meu', 'minha', 'meus', 'minhas',
    'como', 'quero', 'preciso'
]

# ============================================================
# VETORIZAÇÃO TF-IDF COM N-GRAMS
# ============================================================

vectorizer = TfidfVectorizer(
    stop_words=stopwords_pt,
    ngram_range=(1, 2)
)

X_vec = vectorizer.fit_transform(df1['mensagem'])
y = df1['intencao']

# ============================================================
# TREINAMENTO DO MODELO
# ============================================================

modelo = LogisticRegression()
modelo.fit(X_vec, y)

# ============================================================
# TESTE COM UMA NOVA MENSAGEM
# ============================================================

frase_teste = [
    "Preciso urgente da segunda via da fatura"
]

frase_vec = vectorizer.transform(frase_teste)

predicao = modelo.predict(frase_vec)[0]

# ============================================================
# RESULTADOS
# ============================================================

print("--- RESULTADOS DO LAB 01 (AULA 03) ---")

print(f"Mensagem: '{frase_teste[0]}'")

print(f"Intenção Predita: [{predicao}]")

print(
    f"Vocabulário Filtrado (sem stopwords): "
    f"{list(vectorizer.get_feature_names_out())}"
)

print(f"\nQuantidade de palavras no vocabulário: "
      f"{len(vectorizer.get_feature_names_out())}")


# ============================================================
# RELATÓRIO
# ============================================================

print("\n========== RELATÓRIO ==========")

print("""
1 - Qual o impacto da remoção de stopwords no tamanho do vocabulário?

A remoção de stopwords diminui o tamanho do vocabulário utilizado
pelo modelo, pois palavras muito comuns e que possuem pouca
importância para identificar a intenção são retiradas.

Isso ajuda a reduzir a quantidade de informações desnecessárias
e permite que o modelo dê mais importância às palavras relevantes.

Por exemplo, palavras como "de", "da", "o", "a", "como" e "para"
são muito frequentes, mas normalmente não indicam a intenção
principal do usuário.

----------------------------------------

2 - O que significa a configuração ngram_range=(1, 2) no
TfidfVectorizer?

A configuração ngram_range=(1, 2) faz com que o TF-IDF considere
palavras individuais e combinações de duas palavras.

O valor 1 representa os unigramas, que são palavras individuais.

Exemplos:
"boleto"
"fatura"
"dívida"

O valor 2 representa os bigramas, que são combinações de duas
palavras.

Exemplos:
"segunda via"
"fatura atrasada"
"negociar dívida"

Dessa forma, o modelo consegue analisar não somente palavras
isoladas, mas também pequenas combinações de palavras que podem
ter um significado importante para identificar a intenção.

----------------------------------------

3 - Como a remoção de palavras genéricas ajuda a evitar
classificações incorretas?

A remoção de palavras genéricas ajuda o modelo a se concentrar
nas palavras que realmente possuem relação com a intenção do
usuário.

Por exemplo, as palavras "quero", "preciso", "como", "de" e
"minha" aparecem em várias mensagens diferentes.

Se essas palavras fossem consideradas muito importantes, o
modelo poderia encontrar semelhanças entre mensagens que possuem
intenções diferentes.

Ao remover essas palavras, termos mais importantes como
"segunda via", "fatura", "dívida", "acordo", "endereço" e
"cadastro" recebem maior destaque.

Isso melhora a capacidade do modelo de diferenciar as intenções
e pode diminuir classificações incorretas.
""")
```
