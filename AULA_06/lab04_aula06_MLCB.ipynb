```python
def processar_atendimento_sac(
    mensagem_usuario: str
):

    if (
        not mensagem_usuario
        or not mensagem_usuario.strip()
    ):
        return (
            "N/A",
            "0.0%",
            "Por favor, digite uma mensagem para que eu possa ajudar."
        )

    # 1. Pré-processamento da mensagem do usuário
    mensagem_limpa = preprocessar_texto(mensagem_usuario)

    # 2. Extração do embedding da mensagem
    embedding_mensagem = extrair_sentence_embedding(mensagem_limpa, word_vectors)

    # Reshape para o formato esperado pelo modelo (1 amostra, N características)
    embedding_mensagem = embedding_mensagem.reshape(1, -1)

    # 3. Predição da intenção
    intencao_predita = modelo_nlu.predict(embedding_mensagem)[0]

    # 4. Cálculo da confiança
    probabilidades = modelo_nlu.predict_proba(embedding_mensagem)[0]
    indice_intencao = modelo_nlu.classes_.tolist().index(intencao_predita)
    confianca = probabilidades[indice_intencao]

    # 5. Aplicação do limiar de confiança
    if confianca >= LIMIAR_CONFIANCA:
        resposta = RESPOSTAS_PADRAO.get(
            intencao_predita,
            "Desculpe, não encontrei uma resposta padrão para essa intenção."
        )
    else:
        intencao_predita = "indefinida"
        resposta = (
            "Não compreendi a sua solicitação. Por favor, reformule sua pergunta. "
            "Se precisar de ajuda imediata, ligue para nosso SAC: 0800-123-4567."
        )

    return (
        intencao_predita,
        f"{confianca*100:.1f}%",
        resposta
    )
```
