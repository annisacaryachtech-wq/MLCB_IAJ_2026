```python
# Claro - aqui está o código reto, sem explicações e sem divisões desnecessárias, pronto para colar no Colab:

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

dados = {
    "mensagem": [
        "Quero comprar uma passagem para Orlando",
        "Gostaria de reservar um voo para Salvador",
        "Qual o preço da passagem para Lisboa?",
        "Preciso comprar uma passagem para o Rio de Janeiro",
        "Como faço para cancelar minha reserva?",
        "Quero cancelar minha viagem",
        "Gostaria de cancelar minha passagem",
        "Preciso cancelar meu voo",
        "Quero falar com um atendente",
        "Pode me transferir para o suporte?",
        "Preciso falar com uma pessoa do atendimento",
        "Gostaria de conversar com um operador"
    ],
    "intencao": [
        "comprar_passagem",
        "comprar_passagem",
        "comprar_passagem",
        "comprar_passagem",
        "cancelar_reserva",
        "cancelar_reserva",
        "cancelar_reserva",
        "cancelar_reserva",
        "falar_atendente",
        "falar_atendente",
        "falar_atendente",
        "falar_atendente"
    ]
}

df = pd.DataFrame(dados)

X_train, X_test, y_train, y_test = train_test_split(
    df["mensagem"],
    df["intencao"],
    test_size=0.25,
    random_state=42,
    stratify=df["intencao"]
)

vectorizer = TfidfVectorizer()
X_train = vectorizer.fit_transform(X_train)
X_test = vectorizer.transform(X_test)

modelo = LogisticRegression()
modelo.fit(X_train, y_train)

y_pred = modelo.predict(X_test)
print("Acurácia:", accuracy_score(y_test, y_pred))

frases = [
    "Quanto custa uma passagem para Paris?",
    "Quero cancelar o bilhete que comprei",
    "Preciso falar com um atendente humano"
]

frases_vec = vectorizer.transform(frases)
predicoes = modelo.predict(frases_vec)

for frase, predicao in zip(frases, predicoes):
    print("Mensagem:", frase)
    print("Intenção:", predicao)
    print()
```
