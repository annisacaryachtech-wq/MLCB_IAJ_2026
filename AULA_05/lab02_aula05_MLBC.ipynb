## EXERCÍCIO 2 - SENTENCE EMBEDDINGS COM FASTTEXT E MEAN POOLING

```python
import numpy as np
import pandas as pd
import gensim.downloader as api

print("Carregando modelo de Embeddings...")

# Using a FastText model as indicated by the exercise title.
# The 'glove-wiki-gigaword-50' model is a GloVe model, not FastText.
# A common FastText model available via gensim.downloader is 'fasttext-wiki-news-subwords-300'.
# Note: This model is quite large (around 1GB) and might take some time to download.
fasttext_model = api.load("fasttext-wiki-news-subwords-300")

def obter_vetor_frase(frase, model):
    # Divide a frase em palavras
    palavras = frase.split()
    vetores = []

    # Pega o vetor de cada palavra que estiver no modelo
    for palavra in palavras:
        # Ensure the word is lowercase for better matching with pre-trained models
        if palavra.lower() in model:
            vetores.append(model[palavra.lower()])

    # Caso nenhuma palavra seja encontrada
    if len(vetores) == 0:
        return np.zeros(model.vector_size)

    # Faz a média dos vetores das palavras
    vetor_medio = np.mean(vetores, axis=0)

    return vetor_medio


# Carregando o arquivo da atividade
# Make sure 'sac_moveis_ac2.csv' is uploaded to your Colab environment (e.g., /content/sac_moveis_ac2.csv)
df = pd.read_csv("sac_moveis_ac2.csv")

# Transformando cada mensagem em um vetor
X_vetores = np.array([
    obter_vetor_frase(str(msg), fasttext_model) # Convert message to string to handle potential non-string types
    for msg in df["mensagem"]
])

print("Formato da Matriz de Vetores Densos:", X_vetores.shape)
```
